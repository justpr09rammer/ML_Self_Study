# Text to categorical data

Frequency encoding

```python
# These columns have hundreds of unique values with almost no order.
for col in ['event', 'main_speaker', 'speaker_occupation']:
    freq_map = train[col].value_counts().to_dict()
    train[col] = train[col].map(freq_map).fillna(1)
    test[col]  = test[col].map(freq_map).fillna(1)
```

```python
import ast

def safe_parse(s):
    try:
        return ast.literal_eval(s)
    except:
        return []

# Feature 1: how many tags does this talk have?
train['num_tags'] = train['tags'].apply(lambda x: len(safe_parse(x)))
test['num_tags']  = test['tags'].apply(lambda x: len(safe_parse(x)))

# Feature 2: one-hot the TOP 20 most common tags
from collections import Counter

all_tags = [tag for tags in train['tags'].apply(safe_parse) for tag in tags]
top_tags = [tag for tag, _ in Counter(all_tags).most_common(20)]

print("Top 20 tags:", top_tags)

for tag in top_tags:
    train[f'tag_{tag}'] = train['tags'].apply(lambda x: int(tag in safe_parse(x)))
    test[f'tag_{tag}']  = test['tags'].apply(lambda x: int(tag in safe_parse(x)))

train.drop(columns=['tags'], inplace=True)
test.drop(columns=['tags'], inplace=True)
```

Extracting ratings

```python
def parse_ratings(s):
    """Convert ratings string into a dict of {name: count}"""
    result = {}
    try:
        items = ast.literal_eval(s)
        for item in items:
            result[item['name']] = item['count']
    except:
        pass
    return result

# Get all unique rating names

sample_ratings = parse_ratings(train['ratings'].iloc[0])
rating_names = list(sample_ratings.keys())
print("Rating categories:", rating_names)
# ['Obnoxious', 'Unconvincing', 'OK', 'Longwinded', 'Funny',
#  'Fascinating', 'Inspiring', 'Persuasive', 'Informative',
#  'Ingenious', 'Jaw-dropping', 'Confusing', 'Courageous', 'Beautiful']

for df in [train, test]:
    parsed = df['ratings'].apply(parse_ratings)
    for name in rating_names:
        df[f'rating_{name}'] = parsed.apply(lambda x: x.get(name, 0))

    # Also: total rating count (how engaged was the audience overall?)
    df['total_ratings'] = df[[f'rating_{n}' for n in rating_names]].sum(axis=1)

    # Positive sentiment ratio
    positive = ['Inspiring', 'Fascinating', 'Persuasive', 'Informative',
                'Ingenious', 'Jaw-dropping', 'Courageous', 'Beautiful', 'Funny']
    df['positive_rating_ratio'] = (
        df[[f'rating_{n}' for n in positive]].sum(axis=1) / (df['total_ratings'] + 1)
    )

train.drop(columns=['ratings'], inplace=True)
test.drop(columns=['ratings'], inplace=True)
```

```python
#meaning that if a ted talk is popular, similar ted talks to that could have higher chances to be popular as well, so sincce we neet to use numerical values
#we need to get its mean values for realted talks
def avg_related_views(s):
    try:
        items = ast.literal_eval(s)
        counts = [item['viewed_count'] for item in items if 'viewed_count' in item]
        return np.mean(counts) if counts else 0
    except:
        return 0

train['avg_related_views'] = train['related_talks'].apply(avg_related_views)
test['avg_related_views']  = test['related_talks'].apply(avg_related_views)

train.drop(columns=['related_talks'], inplace=True)
test.drop(columns=['related_talks'], inplace=True)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Description: 100 features
tfidf_desc = TfidfVectorizer(max_features=100, stop_words='english', ngram_range=(1,2))
desc_train = tfidf_desc.fit_transform(train['description'].fillna(''))
desc_test  = tfidf_desc.transform(test['description'].fillna(''))

desc_train_df = pd.DataFrame(desc_train.toarray(),
                              columns=[f'desc_{i}' for i in range(100)],
                              index=train.index)
desc_test_df  = pd.DataFrame(desc_test.toarray(),
                              columns=[f'desc_{i}' for i in range(100)],
                              index=test.index)

# Title: 50 features (shorter text, fewer needed)
tfidf_title = TfidfVectorizer(max_features=50, stop_words='english')
title_train = tfidf_title.fit_transform(train['title'].fillna(''))
title_test  = tfidf_title.transform(test['title'].fillna(''))

title_train_df = pd.DataFrame(title_train.toarray(),
                               columns=[f'title_{i}' for i in range(50)],
                               index=train.index)
title_test_df  = pd.DataFrame(title_test.toarray(),
                               columns=[f'title_{i}' for i in range(50)],
                               index=test.index)

train = pd.concat([train.drop(columns=['description', 'title']), desc_train_df, title_train_df], axis=1)
test  = pd.concat([test.drop(columns=['description', 'title']),  desc_test_df,  title_test_df],  axis=1)
```

# Model evaulation

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

target    = np.log1p(train['views'])
X         = train.drop(columns=['views', 'id'])
X_test    = test.drop(columns=['id'])
X_test    = X_test.reindex(columns=X.columns, fill_value=0)

# Train/validation split
X_train, X_val, y_train, y_val = train_test_split(X, target, test_size=0.2, random_state=42)

# Scale (needed for linear models, SVM, KNN)
scaler   = StandardScaler()
Xtr_sc   = scaler.fit_transform(X_train)
Xval_sc  = scaler.transform(X_val)

print(f"Features: {X.shape[1]}  |  Train rows: {X_train.shape[0]}  |  Val rows: {X_val.shape[0]}")
```

```python
from sklearn.metrics import mean_squared_log_error
#scoring helper
def rmsle(model, X_val, y_val, name):
    preds = model.predict(X_val)
    # clamp negatives (linear models can predict negative log-views)
    preds = np.clip(preds, 0, None)
    score = np.sqrt(mean_squared_log_error(np.expm1(y_val), np.expm1(preds)))
    print(f"  {name:<35} RMSLE: {score:.4f}")
    return score

results = {}
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(Xtr_sc, y_train)
results['1. Linear Regression'] = rmsle(lr, Xval_sc, y_val, '1. Linear Regression')

# Which features matter most?
coef_df = pd.Series(lr.coef_, index=X.columns).sort_values(key=abs, ascending=False)
print("\nTop 10 most influential features:")
print(coef_df.head(10))
```
