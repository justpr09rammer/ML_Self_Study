Linear Combination

```python
#Linear combination of vectors
#v = c1v1 + c2v2 + c3v3 + ... + cnvn
#c is a scalar(coefficients), v is a vector
v1 = np.array([1,1,2,3], dtype = float)
v2 = np.array([0,-1,2,-4])
c1 = 3
c2 = 2
print(f"Linear combination of v1 and v2: {c1*v1 + c2*v2}")
```

```python
def lincomb(coef, vectors):
  if (len(coef) != len(vectors)):
      raise ValueError("Number of coefficients must equal number of vectors")
  for v in vectors:
    if len(v) != len(vectors[0]):
        raise ValueError("All vectors must have the same dimension")
  n = len(coef)
  result_vector = np.zeros(len(vectors[0]))
  for i in range(n):
    result_vector += coef[i] * vectors[i]
  return result_vector
```

```python
x = np.random.random(5)
y = np.random.random(5)
z = np.random.random(5)
u = np.random.random(5)
vectors = [x,y,z,u]
coef = [-1,2,3,9]
vectors = [x,y,z,u]
v = lincomb(coef, vectors)
print(f"linear combination of x, y and z with coefficients {coef}: {v}")
```

# Linear Combination of Vectors Visualization in 2D




```python
#v1 and v2 are linearly independent, that spans the whole 2d space,
#but if we take u = v1 + c2*v2, here we will look at if we keep the first vector and add multiples of v2, all new vector end points will be on a signle line
v1 = np.array([10, 40])
v2 = np.array([50, 30])

plt.quiver(0, 0, v1[0], v1[1],
           angles='xy', scale_units='xy', scale=1,
           color='red')

plt.quiver(0, 0, v2[0], v2[1],
           angles='xy', scale_units='xy', scale=1,
           color='green')
xs = []
ys = []
for i in range(5):
    c2 = np.random.randint(-6, 6)
    u = v1 + c2 * v2
    plt.quiver(0, 0, u[0], u[1],
              angles='xy', scale_units='xy', scale=1,
              color='black')
    xs.append(u[0])
    ys.append(u[1])
plt.plot(xs, ys)
plt.scatter(xs, ys)
plt.axhline(0, color='black')
plt.axvline(0, color='black')
plt.grid(True)
plt.gca().set_aspect('equal')
plt.savefig('linear_combination.png')

plt.show()
```

Span of Two Linearly Independent Vectors in 2D Graph

```python
v1 = np.array([10, 40])
v2 = np.array([50, 30])

plt.quiver(0, 0, v1[0], v1[1],
           angles='xy', scale_units='xy', scale=1,
           color='red')

plt.quiver(0, 0, v2[0], v2[1],
           angles='xy', scale_units='xy', scale=1,
           color='green')

x = []
y = []
for c1 in range(-10, 11):
    xs = []
    ys = []
    for c2 in range(-10, 11):
        u = c1 * v1 + c2 * v2
        xs.append(u[0])
        ys.append(u[1])
        x.append(u[0])
        y.append(u[1])

    plt.plot(xs, ys, color=np.random.rand(3))

plt.scatter(x, y, s=50)

plt.axhline(0, color='black')
plt.axvline(0, color='black')
plt.grid(True)

plt.xlim(-200, 200)
plt.ylim(-200, 200)

plt.gca().set_aspect('equal')
plt.savefig('linear_combination_v1.png')
plt.show()
```

# Span of Vectors in 3D Graph

```python
#span of two linearly independent vectors in 3D space, which is a plane showed visually below
v1 = np.array([4, 2, 1])
v2 = np.array([-2, 5, 2])

axis_limit = 40

x_pts, y_pts, z_pts = [], [], []
for c1 in range(-5, 6):
    for c2 in range(-5, 6):
        u = c1 * v1 + c2 * v2
        x_pts.append(u[0])
        y_pts.append(u[1])
        z_pts.append(u[2])

def style_3d_axes(ax, title_text):
    ax.set_xlabel('X Axis (Left)')
    ax.set_ylabel('Y Axis')
    ax.set_zlabel('Z Axis')
    ax.set_xlim([-axis_limit, axis_limit])
    ax.set_ylim([-axis_limit, axis_limit])
    ax.set_zlim([-axis_limit, axis_limit])

    # Standard center axis line rails passing through (0,0,0)
    ax.plot([-axis_limit, axis_limit], [0, 0], [0, 0], color='black', linewidth=1.5)
    ax.plot([0, 0], [-axis_limit, axis_limit], [0, 0], color='black', linewidth=1.5)
    ax.plot([0, 0], [0, 0], [-axis_limit, axis_limit], color='black', linewidth=1.5)

    # Apply your custom camera angle and aspect ratio fixes
    ax.view_init(elev=20, azim=-135)
    ax.set_box_aspect([1, 1, 1])
    ax.set_title(title_text, fontsize=12, pad=15)

fig1 = plt.figure(figsize=(8, 8))
ax1 = fig1.add_subplot(projection='3d')

ax1.quiver(0, 0, 0, v1[0], v1[1], v1[2], color='red', linewidth=3, arrow_length_ratio=0.15, label='v1')
ax1.quiver(0, 0, 0, v2[0], v2[1], v2[2], color='green', linewidth=3, arrow_length_ratio=0.15, label='v2')

ax1.scatter(x_pts, y_pts, z_pts, color='royalblue', s=30, alpha=0.7, label='c1*v1 + c2*v2')

style_3d_axes(ax1, "Step 1: Scatter Cloud of Linear Combinations\n(Notice how the dots form a flat sheet in 3D space)")
ax1.legend(loc="upper left")
fig1.savefig('step1_scatter_cloud.png', dpi=300)
fig2 = plt.figure(figsize=(10, 8))
ax2 = fig2.add_subplot(projection='3d')

ax2.quiver(0, 0, 0, v1[0], v1[1], v1[2], color='red', linewidth=4, arrow_length_ratio=0.15, label='v1', zorder=5)
ax2.quiver(0, 0, 0, v2[0], v2[1], v2[2], color='green', linewidth=4, arrow_length_ratio=0.15, label='v2', zorder=5)

colors = cm.rainbow(np.linspace(0, 1, 11))

for i, c1 in enumerate(range(-5, 6)):
    line_color = colors[i]

    line_x, line_y, line_z = [], [], []
    for c2 in range(-5, 6):
        u = c1 * v1 + c2 * v2
        line_x.append(u[0])
        line_y.append(u[1])
        line_z.append(u[2])

    ax2.plot(line_x, line_y, line_z, color=line_color, linewidth=2, alpha=0.8,
             label=f'c1 = {c1}' if i % 2 == 0 else "")

ax2.scatter(x_pts, y_pts, z_pts, color='black', s=20, alpha=0.4, zorder=3)

c1_grid, c2_grid = np.meshgrid(np.linspace(-5, 5, 10), np.linspace(-5, 5, 10))
surf_X = c1_grid * v1[0] + c2_grid * v2[0]
surf_Y = c1_grid * v1[1] + c2_grid * v2[1]
surf_Z = c1_grid * v1[2] + c2_grid * v2[2]
ax2.plot_surface(surf_X, surf_Y, surf_Z, color='gray', alpha=0.1, zorder=1)

style_3d_axes(ax2, "Step 2: Vector Span Lines Grouped by $c_1$ Coefficient\n(Each tracking color follows a distinct $v_1$ translation step)")

ax2.legend(loc="upper left", bbox_to_anchor=(1.05, 1), title="Coefficients")

plt.tight_layout()
fig2.savefig('step2_color_coded_span.png', dpi=300)

plt.show()
```

```python
#Span of 3 linearly independent vectors in 3D graph which spans the whole space since its rank is 3

v1 = np.array([12, 4, 2])
v2 = np.array([-4, 15, 3])
v3 = np.array([2, -2, 14])

axis_limit = 60

x_pts, y_pts, z_pts = [], [], []
for c1 in range(-3, 4):
    for c2 in range(-3, 4):
        for c3 in range(-3, 4):
            u = c1 * v1 + c2 * v2 + c3 * v3
            x_pts.append(u[0])
            y_pts.append(u[1])
            z_pts.append(u[2])

def style_3d_axes(ax, title_text):
    ax.set_xlabel('X Axis (Left)')
    ax.set_ylabel('Y Axis')
    ax.set_zlabel('Z Axis')
    ax.set_xlim([-axis_limit, axis_limit])
    ax.set_ylim([-axis_limit, axis_limit])
    ax.set_zlim([-axis_limit, axis_limit])
    ax.plot([-axis_limit, axis_limit], [0, 0], [0, 0], color='black', linewidth=1.5)
    ax.plot([0, 0], [-axis_limit, axis_limit], [0, 0], color='black', linewidth=1.5)
    ax.plot([0, 0], [0, 0], [-axis_limit, axis_limit], color='black', linewidth=1.5)
    ax.view_init(elev=20, azim=-135)
    ax.set_box_aspect([1, 1, 1])
    ax.set_title(title_text, fontsize=12, pad=15)

fig1 = plt.figure(figsize=(8, 8))
ax1 = fig1.add_subplot(projection='3d')
ax1.quiver(0, 0, 0, v1[0], v1[1], v1[2], color='red', linewidth=3, arrow_length_ratio=0.15, label='v1')
ax1.quiver(0, 0, 0, v2[0], v2[1], v2[2], color='green', linewidth=3, arrow_length_ratio=0.15, label='v2')
ax1.quiver(0, 0, 0, v3[0], v3[1], v3[2], color='orange', linewidth=3, arrow_length_ratio=0.15, label='v3')
ax1.scatter(x_pts, y_pts, z_pts, color='royalblue', s=20, alpha=0.6, label='c1*v1 + c2*v2 + c3*v3')
style_3d_axes(ax1, "Step 1: Scatter Cloud of 3D Linear Combinations\n(Notice how the dots fill the entire 3D volume)")
ax1.legend(loc="upper left")
fig1.savefig('step1_3d_scatter_cloud.png', dpi=300)

fig2 = plt.figure(figsize=(10, 8))
ax2 = fig2.add_subplot(projection='3d')
ax2.quiver(0, 0, 0, v1[0], v1[1], v1[2], color='red', linewidth=4, arrow_length_ratio=0.15, label='v1', zorder=5)
ax2.quiver(0, 0, 0, v2[0], v2[1], v2[2], color='green', linewidth=4, arrow_length_ratio=0.15, label='v2', zorder=5)
ax2.quiver(0, 0, 0, v3[0], v3[1], v3[2], color='orange', linewidth=4, arrow_length_ratio=0.15, label='v3', zorder=5)

colors = cm.rainbow(np.linspace(0, 1, 7))

for i, c1 in enumerate(range(-3, 4)):
    line_color = colors[i]
    for c2 in range(-3, 4):
        line_x, line_y, line_z = [], [], []
        for c3 in range(-3, 4):
            u = c1 * v1 + c2 * v2 + c3 * v3
            line_x.append(u[0])
            line_y.append(u[1])
            line_z.append(u[2])
        ax2.plot(line_x, line_y, line_z, color=line_color, linewidth=1.5, alpha=0.7,
                 label=f'c1 = {c1}' if c2 == 0 and i % 2 == 0 else "")

ax2.scatter(x_pts, y_pts, z_pts, color='black', s=10, alpha=0.3, zorder=3)
style_3d_axes(ax2, "Step 2: 3D Vector Space Grid Lines Grouped by $c_1$\n(The full 3D coordinate lattice framework)")
ax2.legend(loc="upper left", bbox_to_anchor=(1.05, 1), title="Coefficients")
plt.tight_layout()
fig2.savefig('step2_3d_color_coded_span.png', dpi=300)

plt.show()
```
